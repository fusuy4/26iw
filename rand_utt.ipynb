```python
# For reading in random utterances from a dataframe, 
# (done for earlier runs where sampling wasn't being used)
import csv
import pandas as pd

df_M_utt = pd.read_csv("M/data_sentiment/Doom_utt_final.csv")
df_T_utt = pd.read_csv("T/Data/destiny2_utt_sentiment_final.csv")
df_E_utt = pd.read_csv("E_10+/Data/PvZGardenWarfare_sent_utt.csv")

df_M_convo = pd.read_csv("M/data_sentiment/Doom_convo_final.csv")
df_T_convo = pd.read_csv("T/Data/destiny2_convo_sentiment_final.csv")
df_E_convo = pd.read_csv("E_10+/Data/PvZGardenWarfare_convo_sent_final.csv")

list_of_corpora = [df_M_utt, df_T_utt, df_E_utt]

sampled_conversations = df_M_convo["conversation_id"].sample(
    n=12500, 
    random_state=42 
)

filt_utterances = df_M_utt[df_M_utt["conversation_id"].isin(sampled_conversations)]

filt_utterances.to_csv("Doom_utt_final_filt.csv", quoting=csv.QUOTE_ALL) 

filtered_conversations = df_M_convo[df_M_convo["conversation_id"].isin(sampled_conversations)]

filtered_conversations.to_csv("Doom_convo_final_filt.csv", quoting=csv.QUOTE_ALL)
```
