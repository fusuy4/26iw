```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
from nltk import word_tokenize
```

```python
df_E_utt = pd.read_csv("combined_utt_sent_E.csv")
df_T_utt = pd.read_csv("combined_utt_sent_T.csv")
df_M_utt = pd.read_csv("combined_utt_sent_M.csv")
```

```python
def tokenize(text):
    tokenizedText = word_tokenize(text)

    return tokenizedText
```

```python
print(tokenize("MY name is jeffery djsdnjksnvcs  dd''dfv;  dd"))
```

### Posts, Convo Lenths

```python
def compute_corpus_length_stats(df):
    df = df.copy()
    
    df["tokens"] = df["text"].fillna("").apply(lambda x: tokenize(x))
    
    df["utterance_length"] = df["tokens"].apply(len)
    
    avg_utt_length = df["utterance_length"].mean()

    conv_lengths = df.groupby("conversation_id")["utterance_id"].count()
    conv_word_lengths = df.groupby("conversation_id")["utterance_length"].sum()
    
    stats_df = pd.DataFrame({
        "utterance_length": df["utterance_length"].describe(),
        "conv_len_utterances": conv_lengths.describe(),
        "conv_len_words": conv_word_lengths.describe()
    })
    
    
    return stats_df
```

```python
len_stats_M_df = compute_corpus_length_stats(df_M_utt)
len_stats_T_df = compute_corpus_length_stats(df_T_utt)
len_stats_E_df = compute_corpus_length_stats(df_E_utt)
```
