```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df_M_convo_topic = pd.read_csv("M/data_topic/final_topic_convo_M.csv")
df_T_convo_topic = pd.read_csv("T/data_topics/final_topic_convo_T.csv")
df_E_convo_topic = pd.read_csv("E_10+/final_topic_convo_E.csv")
```

```python
counts_M = df_M_convo_topic["topic_label"].value_counts()
counts_M_normalized = df_M_convo_topic["topic_label"].value_counts(normalize=True) * 100

print(counts_M)

print(counts_M_normalized)

print(counts_M_normalized[1:])

counts_M.plot(kind="bar")
```

```python
counts_T = df_T_convo_topic["topic_label"].value_counts()
counts_T_normalized = df_T_convo_topic["topic_label"].value_counts(normalize=True)

print(counts_T)

print(counts_T_normalized)

print(counts_T_normalized[1:])

counts_T.plot(kind="bar")
```

```python
counts_E = df_E_convo_topic["topic_label"].value_counts()
counts_E_normalized = df_E_convo_topic["topic_label"].value_counts(normalize=True)

print(counts_E)

print(counts_E_normalized)

print(counts_E_normalized[1:])

counts_E.plot(kind="bar")
```
